# Teste do Modelo 3: TabPFN-TS

```python
import pandas as pd
import torch
import os

# --- 1. Configurações ---
DATA_DIR = "../../data"
HORIZONTE_PREVISAO = 14
N_DIAS_HISTORICO = 100

# --- 2. Carregar Dados ---
print("Carregando dados...")
hist_path = os.path.join(DATA_DIR, "hist.parquet")
df_hist = pd.read_parquet(hist_path)
print("Dados históricos carregados.")

# --- 3. Definir dispositivo ---
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"\nUsando dispositivo: {device}")
```

```python
# ==============================================================================
# TESTE 3: TabPFN-TS
# ==============================================================================
# Covariáveis: Não. Modelo univariado.
# ==============================================================================

try:
    from tabpfn_time_series import TabPFNTimeSeriesRegressor
    import warnings

    # TabPFN pode gerar muitos avisos
    warnings.filterwarnings('ignore', category=UserWarning)

    print("\n--- Testando TabPFN-TS ---")

    # 1. Carregar o modelo
    model = TabPFNTimeSeriesRegressor(
        device=device,
        N_ensemble_configurations=4,
        context_length=min(N_DIAS_HISTORICO, 128)
    )

    # 2. Preparar dados
    y_hist = df_hist['target'].values

    # 3. Treinar (Fit) e Prever (Predict)
    print("Rodando 'fit'...")
    model.fit(y_hist)

    print(f"Rodando 'predict' para {HORIZONTE_PREVISAO} passos...")
    y_pred = model.predict(HORIZONTE_PREVISAO)

    print("Previsão (primeiros 5 valores):", y_pred.round(2)[:5])
    print("Teste do TabPFN-TS concluído.\n")

except ImportError:
    print("TabPFN-TS não instalado. Pulando teste.")
except Exception as e:
    print(f"Erro ao rodar TabPFN-TS: {e}")
```
